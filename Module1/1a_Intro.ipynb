## 2. Numpy Arrays

### Images are arrays of numbers

[Numpy](https://numpy.org) is a library for processing multi-dimensional lists of numbers, of which microscopy images (stacks, multi-channel, time-lapses etc.) are a prominent example.

An ndarray is a (usually fixed-size) multidimensional container of items of the same type and size. The number of dimensions and items in an array is defined by its shape , which is a tuple of N non-negative integers that specify the sizes of each dimension.

*Source: [Indexing on ndarrays](https://numpy.org/doc/stable/user/basics.indexing.html) </br>
See also : [Image manipulation and processing using Numpy and Scipy by Emmanuelle Gouillart and Gaël Varoquaux](https://scipy-lectures.org/advanced/image_processing/index.html#basic-image)*

### Different ways of creating arrays

There are many different ways of creating arrays, we see only the most useful for us:

- ```np.array``` - basic creation of the array directly defnining itself 

- ```np.arange``` - reates arrays with regularly incrementing values.

- ```np.zeros``` - create an image where all pixels have value `0`, it requires to specify and image size.

- ```np.random.random``` - create an image with random values, it requires to specify and image size.

*For more: [Array creation](https://numpy.org/doc/stable/user/basics.creation.html)*

**Two important properties**:

- the ```shape``` of the array, i.e. the number of rows, columns (and channels, planes etc. for multi-dimensional images)
- the ```dtype``` of the array, i.e. an image of type `int64` has 2 to the power of 64 different grey values.

````python
import numpy as np

x = np.array([[1, 2, 3], [4, 5, 6]], np.int32)
print(type(x))
print(x.shape)
print(x.dtype)
````

An image is just a two dimensional list of pixels values, in other words a matrix, with a certain number of rows and columns. Therefore we can define it as a list of lists, each list being a row of pixels:

````python
raw_image_array = np.asarray([
    [1, 0, 2, 1, 0, 0, 0],
    [0, 3, 1, 0, 1, 0, 1],
    [0, 5, 5, 1, 0, 1, 0],
    [0, 6, 6, 5, 1, 0, 2],
    [0, 0, 5, 6, 3, 0, 1],
    [0, 1, 2, 1, 0, 0, 1],
    [1, 0, 1, 0, 0, 1, 0]
])
print(type(raw_image_array))
print(raw_image_array.shape)
print(raw_image_array.dtype)
print(raw_image_array)
````

Such Numpy arrays we can perform the computations, such as:

````python
raw_image_array - 2
````

## 3. Opening and visualizing images

Most images with standard extensions (tif, png etc.) can be read using the [```skimage.io.imread```](https://scikit-image.org/docs/dev/api/skimage.io.html#skimage.io.imread) function. 

See also
* [Opening other file formats using Python](https://haesleinhuepf.github.io/BioImageAnalysisNotebooks/12a_image_file_formats/readme.html)

To use ```imread``` you have three possibilities:
- use the absolute path to an image file e.g. ```imread('/Users/username/Desktop/blobs.tif')```
- use a relative path to where you currently are, e.g. ```imread('data/blobs.tif')``` or ```imread('../../data/blobs.tif')```
- use a url that points to an image file, e.g. from a GitHub repository ```imread('https://github.com/haesleinhuepf/BioImageAnalysisNotebooks/raw/main/data/blobs.tif')```

We will use this and see how it works in the next section.

## 4. Displaying Images: Matplotlib

```Matplotlib.pyplot``` is a collection of functions that make matplotlib work like MATLAB, you can use it to create your graphs or to visualize your images.

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(2,2))
plt.plot([1, 2, 3, 4])

plt.figure(figsize=(2,2))
#plot([x],[y])
plt.plot([1, 4, 6, 8], [1, 2, 3, 4]) #, 'ro') 
plt.ylabel('y')
plt.xlabel('x')
plt.grid()
plt.show()
````

````python
names = ['group_a', 'group_b', 'group_c']
values = [1, 10, 100]

plt.figure(figsize=(9, 3))

plt.subplot(131)
plt.bar(names, values)

plt.subplot(132)
plt.scatter(names, values)

plt.subplot(133)
plt.plot(names, values)

plt.suptitle('Categorical Plotting')
plt.show()
````

There are many ways to display simple 2D images, but the most common one is Matplotlib.pyplot's `imshow` function.

The most important plotting options for our purposes are
* `cmap` to change the colormap (LUT)
* `vmin` to change the pixel value corresponding to the first color in the colormap
* `vmax` to change the pixel value corresponding to the last color in the colormap

The last two options control the brightness/contrast.

````python
from skimage import io
cell = io.imread('./data/cell.jpg')

fig, (ax1, ax2, ax3, ax4) = plt.subplots(1, 4, figsize=(14,4))
ax1.imshow(cell, cmap='gray')
ax2.imshow(cell, cmap='gray', vmin=0, vmax=100)
ax3.imshow(cell, cmap='gray', vmin=100, vmax=255)
ax4.imshow(cell, cmap="seismic")
plt.show()
````

*Notes*

> **Parentheses**
> 
> It may initially be confusing why we sometimes need parentheses `()` and sometimes we don't, e.g. `im.flatten()` vs. `im.shape`.
> 
> As a general rule, parentheses indicate that we're calling a method that *does* something (e.g. prints a value, calculates an average, flattens an array).
> 
> When we don't see parentheses, this indicates that we're accessing a field or property (e.g. the shape of an array).
> 
> In practice, the distinction can sometimes be a bit murky as you get deeper into Python - but it helps as a general guide.
